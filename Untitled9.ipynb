```python
import numpy as np
import pandas as pd
import urllib.request
import zipfile
import os
np.random.seed(42)
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
import joblib
from sklearn.model_selection import StratifiedKFold, cross_val_score

# دانلود و استخراج مجموعه داده از مخزن UCI
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00320/student.zip"
urllib.request.urlretrieve(url, "student.zip")

with zipfile.ZipFile("student.zip", "r") as zip_ref:
    zip_ref.extractall("student_data")

# خواندن فایل اطلاعات ریاضیات (student-mat.csv) با جداکننده نقطه ویرگول
df = pd.read_csv("student_data/student-mat.csv", sep=";")
print("ابعاد مجموعه داده:", df.shape)
```

```python
# تبدیل ستون‌های متنی به مقادیر عددی
for col in df.select_dtypes(include=['object']).columns:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])

# سناریوی اول: طبقه‌بندی دوکلاسه بر اساس نمره G3
df['G3_binary'] = np.where(df['G3'] < 10, 0, 1)

# تفکیک ویژگی‌ها از متغیر هدف    
X = df.drop(columns=['G3', 'G3_binary'])
y = df['G3_binary']
```

```python
# سناریوی دوم: طبقه‌بندی پنج‌کلاسه     
def classify_5_classes(grade):
    if grade <= 9:
        return 0
    elif grade <= 11:
        return 1
    elif grade <= 13:
        return 2
    elif grade <= 15:
        return 3
    else:
        return 4

df['G3_5class'] = df['G3'].apply(classify_5_classes)

# تفکیک ویژگی‌ها از متغیر هدف (حذف G3 و ستون‌های  )
X = df.drop(columns=['G3', 'G3_binary', 'G3_5class'])
y_binary = df['G3_binary']
y_5class = df['G3_5class']

cv = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.model_selection import cross_validate
from sklearn.metrics import confusion_matrix

# تعریف مدل‌ها    
models = {
    'KNN': Pipeline([
        ('scaler', StandardScaler()),
        ('classifier', KNeighborsClassifier(n_neighbors=5))
    ]),
    'SVM': Pipeline([
        ('scaler', StandardScaler()),
        ('classifier', SVC(kernel='rbf', probability=True, random_state=42))
    ]),
    'Random Forest': Pipeline([
        ('classifier', RandomForestClassifier(n_estimators=50, random_state=42))
    ])
}


# ارزیابی سناریوی دوکلاسه 

print("--- نتایج ارزیابی سناریوی دوکلاسه (Pass/Fail) ---")
scoring_binary = {
    'accuracy': 'accuracy',
    'precision': 'precision',
    'recall': 'recall',
    'f1': 'f1',
    'roc_auc': 'roc_auc'
}

for name, model in models.items():
    scores = cross_validate(model, X, y_binary, cv=cv, scoring=scoring_binary)
    print(f"مدل: {name:15}")
    print(f"  - Accuracy:  {scores['test_accuracy'].mean():.4f}")
    print(f"  - Precision: {scores['test_precision'].mean():.4f}")
    print(f"  - Recall:    {scores['test_recall'].mean():.4f}")
    print(f"  - F1-Score:  {scores['test_f1'].mean():.4f}")
    print(f"  - AUC:       {scores['test_roc_auc'].mean():.4f}")
    print("-" * 50)
```

```python
# ارزیابی سناریوی پنج کلاسه 
print("\n--- نتایج ارزیابی سناریوی پنج‌کلاسه (5-Class) ---")
scoring_multiclass = {
    'accuracy': 'accuracy',
    'precision': 'precision_macro',
    'recall': 'recall_macro',
    'f1': 'f1_macro',
    'roc_auc': 'roc_auc_ovr'
}

for name, model in models.items():
    scores = cross_validate(model, X, y_5class, cv=cv, scoring=scoring_multiclass)
    print(f"مدل: {name:15}")
    print(f"  - Accuracy:  {scores['test_accuracy'].mean():.4f}")
    print(f"  - Precision: {scores['test_precision'].mean():.4f}")
    print(f"  - Recall:    {scores['test_recall'].mean():.4f}")
    print(f"  - F1-Score:  {scores['test_f1'].mean():.4f}")
    print(f"  - AUC:       {scores['test_roc_auc'].mean():.4f}")
    print("-" * 50)
```

```python
import joblib

# آموزش نهایی مدل برتر (Random Forest) روی کل داده‌های دوکلاسه
best_model = RandomForestClassifier(n_estimators=50, random_state=42)
best_model.fit(X, y_binary)

joblib.dump(best_model, 'student_performance_model.pkl')
joblib.dump(list(X.columns), 'model_features.pkl')

```
